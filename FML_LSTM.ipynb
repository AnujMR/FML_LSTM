# **Using Moses Scripts for Tokenization**

```python
tok = MosesTokenizer(lang='en')
tokenized_src = []
tokenized_tgt = []

for ex in subset:
    src = ex["translation"]["en"]
    tgt = ex["translation"]["fr"]

    src_tok = tok.tokenize(src, escape=True)
    tgt_tok = tok.tokenize(tgt, escape=True)

    tokenized_src.append(src_tok)
    tokenized_tgt.append(tgt_tok)
```

```python
tokenized_src = [s + ["<eos>"] for s in tokenized_src]
tokenized_tgt = [t + ["<eos>"] for t in tokenized_tgt]
```

```python
print(tokenized_src[:10])
print(tokenized_tgt[:10])
```

# **Reversing the input strings**

```python
reversed_src = [list(reversed(s)) for s in tokenized_src]
```

```python
print(reversed_src[:10])
```

# **Creating a vocab of most frequent words**

```python
from collections import Counter
special_tokens = ["<pad>", "<unk>", "<eos>"]

counter = Counter()

# Counting frequencies from data
for s,t in zip(reversed_src, tokenized_tgt):
    counter.update([w for w in s if w not in special_tokens])
    counter.update([w for w in t if w not in special_tokens])
```

```python
vocab_size = 40000
most_common = counter.most_common(vocab_size - 3)
```

```python
vocab = ["<pad>", "<unk>", "<eos>"] + [w for w, _ in most_common]
word2id = {w:i for i,w in enumerate(vocab)}
```

```python
with open("vocab.txt", "w") as f:
    for w in vocab:
        f.write(w + "\n")
```

# **Converting tokens into IDs**

```python
def map_to_ids(tokens):
    return [word2id.get(tok, word2id["<unk>"]) for tok in tokens]
```

```python
src_ids = [map_to_ids(s) for s in reversed_src]
tgt_ids = [map_to_ids(t) for t in tokenized_tgt]
```

# **Filtering sentences by maximum allowed length i.e. 50**

```python
MAX_LEN = 50
filtered_src = []
filtered_tgt = []

for s, t in zip(src_ids, tgt_ids):
    if len(s) <= MAX_LEN and len(t) <= MAX_LEN:
        filtered_src.append(s)
        filtered_tgt.append(t)
```

# **Splitting the data into Train(90%), Validation(3%) and Test(7%) set**

```python
import random
random.seed(42)

pairs = list(zip(filtered_src, filtered_tgt))
random.shuffle(pairs)

N = len(pairs)
train_end = int(0.90 * N)
val_end   = int(0.93 * N)

train = pairs[:train_end]
val   = pairs[train_end:val_end]
test  = pairs[val_end:]
```

# **Saving the IDs into files**

```python
def save_ids(pairs, src_path, tgt_path):
    with open(src_path, "w") as fs, open(tgt_path, "w") as ft:
        for s,t in pairs:
            fs.write(" ".join(map(str, s)) + "\n")
            ft.write(" ".join(map(str, t)) + "\n")

save_ids(train, "train.src", "train.tgt")
save_ids(val,   "val.src",   "val.tgt")
save_ids(test,  "test.src",  "test.tgt")
```

```python
import torch
import nltk
from nltk.translate.bleu_score import corpus_bleu
```

```python
nltk.download('punkt')

from model import Encoder, Decoder, greedy_decode, load_vocab, read_id_file  # from your uploaded file

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

word2idx, idx2word = load_vocab("vocab.txt")
VOCAB_SIZE = len(idx2word)
PAD = 0
EOS = 2

checkpoint = torch.load("final_model.pt", map_location=device)

EMBED_SIZE = 256
HIDDEN_SIZE = 256
NUM_LAYERS = 2

encoder = Encoder(VOCAB_SIZE, EMBED_SIZE, HIDDEN_SIZE, NUM_LAYERS).to(device)
decoder = Decoder(VOCAB_SIZE, EMBED_SIZE, HIDDEN_SIZE, NUM_LAYERS).to(device)


encoder.load_state_dict(checkpoint["encoder"])
decoder.load_state_dict(checkpoint["decoder"])
encoder.eval()
decoder.eval()

print("Model restored successfully!")

test_src = read_id_file("test.src")
test_tgt = read_id_file("test.tgt")

def translate(ids):
    src = torch.tensor(ids, dtype=torch.long).unsqueeze(0).to(device)
    src_mask = (src != PAD).float()
    out = greedy_decode(encoder, decoder, src, src_mask, max_len=50)[0]
    return out

references = []
hypotheses = []

for ref_ids, src_ids in zip(test_tgt, test_src):
    pred = translate(src_ids)

    # removing EOS and PAD
    pred = [t for t in pred if t not in (EOS, PAD)]
    ref_ids = [t for t in ref_ids if t not in (EOS, PAD)]

    references.append([ [idx2word[t] for t in ref_ids] ])
    hypotheses.append([idx2word[t] for t in pred])

bleu = corpus_bleu(references, hypotheses) * 100
print(f"\nBLEU Score: {bleu:.2f}")
```
