# **Encoding text into SentencePiece IDs**

```python
%%writefile preprocess_wmt.py
import sentencepiece as spm
from tqdm import tqdm
import argparse

def encode_file(sp, inp, out):
    with open(inp, "r", encoding="utf8") as f:
        lines = f.read().splitlines()

    encoded = []
    for line in tqdm(lines, desc=f"Encoding {inp}"):
        ids = sp.encode(line, out_type=int)
        ids.append(sp.eos_id())
        encoded.append(" ".join(map(str, ids)))

    with open(out, "w", encoding="utf8") as f:
        f.write("\n".join(encoded))


def main():
    parser = argparse.ArgumentParser()
    parser.add_argument("--model", default="bpe.model")
    parser.add_argument("--train_en", default="train.en")
    parser.add_argument("--train_fr", default="train.fr")
    parser.add_argument("--val_en", default="val.en")
    parser.add_argument("--val_fr", default="val.fr")
    parser.add_argument("--test_en", default="test.en")
    parser.add_argument("--test_fr", default="test.fr")
    args = parser.parse_args()

    sp = spm.SentencePieceProcessor(model_file=args.model)

    encode_file(sp, args.train_en, "train.src")
    encode_file(sp, args.train_fr, "train.tgt")

    encode_file(sp, args.val_en, "val.src")
    encode_file(sp, args.val_fr, "val.tgt")

    encode_file(sp, args.test_en, "test.src")
    encode_file(sp, args.test_fr, "test.tgt")

    print("\nPreprocessing complete!")
    print("Files generated:")
    print("train.src, train.tgt")
    print("val.src, val.tgt")
    print("test.src, test.tgt")


if __name__ == "__main__":
    main()
```

# **Encoder-Decoder**

```python
%%writefile model.py
import torch
import torch.nn as nn
import torch.utils.checkpoint as checkpoint

PAD_ID = 0
UNK_ID = 1
BOS_ID = 2
EOS_ID = 3

class Encoder(nn.Module):
    def __init__(self, vocab, dim=256, num_layers=4, dropout=0.3, use_ckpt=False):
        super().__init__()
        self.use_ckpt = use_ckpt

        self.embedding = nn.Embedding(vocab, dim, padding_idx=PAD_ID)
        self.dropout = nn.Dropout(dropout)

        self.lstm = nn.LSTM(
            dim,
            dim,
            num_layers=num_layers,
            dropout=dropout,
            batch_first=True,
            bidirectional=False
        )

    def forward(self, x, mask):
        emb = self.dropout(self.embedding(x))

        if self.use_ckpt:
            out, (h, c) = checkpoint.checkpoint(self.lstm, emb, use_reentrant=False)
        else:
            out, (h, c) = self.lstm(emb)

        return h, c


class Decoder(nn.Module):
    def __init__(self, vocab, dim=256, num_layers=4, dropout=0.3, use_ckpt=False):
        super().__init__()
        self.use_ckpt = use_ckpt

        self.embedding = nn.Embedding(vocab, dim, padding_idx=PAD_ID)
        self.dropout = nn.Dropout(dropout)

        self.lstm = nn.LSTM(
            dim,
            dim,
            num_layers=num_layers,
            dropout=dropout,
            batch_first=True
        )

        self.output = nn.Linear(dim, vocab)

        self.output.weight = self.embedding.weight

    def forward(self, y, hc):
        emb = self.dropout(self.embedding(y))

        if self.use_ckpt:
            out, hc = checkpoint.checkpoint(self.lstm, emb, hc, use_reentrant=False)
        else:
            out, hc = self.lstm(emb, hc)

        logits = self.output(out)
        return logits, hc
```

# **Training**

```python
%%writefile train.py
import torch
import torch.nn.functional as F
from torch.utils.data import Dataset, DataLoader
from model import Encoder, Decoder, PAD_ID
from tqdm import tqdm
import argparse

torch.backends.cuda.matmul.allow_tf32 = True
torch.backends.cudnn.allow_tf32 = True

class SeqDataset(Dataset):
    def __init__(self, src, tgt):
        self.src = open(src).read().splitlines()
        self.tgt = open(tgt).read().splitlines()

    def __len__(self):
        return len(self.src)

    def __getitem__(self, idx):
        s = list(map(int, self.src[idx].split()))
        t = list(map(int, self.tgt[idx].split()))
        return torch.tensor(s), torch.tensor(t)


def collate(batch):
    src, tgt = zip(*batch)
    src = torch.nn.utils.rnn.pad_sequence(src, batch_first=True, padding_value=PAD_ID)
    tgt = torch.nn.utils.rnn.pad_sequence(tgt, batch_first=True, padding_value=PAD_ID)
    mask = (src != PAD_ID).float()
    return src, mask, tgt


def train(args):
    DEVICE = "cuda" if torch.cuda.is_available() else "cpu"
    VOCAB = args.vocab

    train_ds = SeqDataset(args.train_src, args.train_tgt)
    val_ds   = SeqDataset(args.val_src, args.val_tgt)

    train_loader = DataLoader(train_ds, batch_size=args.batch_size,
                              shuffle=True, collate_fn=collate, pin_memory=True)
    val_loader   = DataLoader(val_ds, batch_size=args.batch_size,
                              shuffle=False, collate_fn=collate, pin_memory=True)

    enc = Encoder(VOCAB).to(DEVICE)
    dec = Decoder(VOCAB).to(DEVICE)

    params = list(enc.parameters()) + list(dec.parameters())
    opt = torch.optim.AdamW(params, lr=args.lr, weight_decay=0.01)

    scaler = torch.amp.GradScaler("cuda")

    print(f"\n=== Training on {DEVICE} ===")
    print(f"Batch size = {args.batch_size}")
    print(f"Epochs     = {args.epochs}")
    print(f"Vocab      = {VOCAB}\n")

    for epoch in range(args.epochs):
        enc.train(); dec.train()
        pbar = tqdm(train_loader)

        for src, mask, tgt in pbar:
            src, mask, tgt = src.to(DEVICE), mask.to(DEVICE), tgt.to(DEVICE)

            opt.zero_grad(set_to_none=True)

            inp = tgt[:, :-1]    # Decoder input
            out = tgt[:, 1:]     # Expected output

            with torch.amp.autocast("cuda", dtype=torch.bfloat16):
                h, c = enc(src, mask)
                logits, _ = dec(inp, (h, c))

                loss = F.cross_entropy(
                    logits.reshape(-1, VOCAB),
                    out.reshape(-1),
                    ignore_index=PAD_ID,
                    label_smoothing=0.1
                )

            scaler.scale(loss).backward()
            torch.nn.utils.clip_grad_norm_(params, 1.0)
            scaler.step(opt)
            scaler.update()

            pbar.set_description(f"epoch {epoch+1} loss={loss.item():.4f}")

        # Validation
        enc.eval(); dec.eval()
        total = 0; count = 0

        with torch.no_grad(), torch.amp.autocast("cuda", dtype=torch.bfloat16):
            for src, mask, tgt in val_loader:
                src, mask, tgt = src.to(DEVICE), mask.to(DEVICE), tgt.to(DEVICE)

                inp = tgt[:, :-1]
                out = tgt[:, 1:]

                h, c = enc(src, mask)
                logits, _ = dec(inp, (h, c))

                loss = F.cross_entropy(
                    logits.reshape(-1, VOCAB),
                    out.reshape(-1),
                    ignore_index=PAD_ID
                )
                total += loss.item()
                count += 1

        print(f"\nVAL LOSS epoch {epoch+1}: {total/count:.4f}")

        torch.save(
            {"enc": enc.state_dict(), "dec": dec.state_dict()},
            f"checkpoint_epoch{epoch+1}.pt"
        )


if __name__ == "__main__":
    parser = argparse.ArgumentParser()
    parser.add_argument("--train_src", default="train.src")
    parser.add_argument("--train_tgt", default="train.tgt")
    parser.add_argument("--val_src", default="val.src")
    parser.add_argument("--val_tgt", default="val.tgt")
    parser.add_argument("--vocab", type=int, default=16000)
    parser.add_argument("--batch_size", type=int, default=32)
    parser.add_argument("--epochs", type=int, default=5)
    parser.add_argument("--lr", type=float, default=3e-4)

    args = parser.parse_args()
    train(args)
```

# **evaluate_bleu.py**

```python
%%writefile evaluate_bleu.py
import torch
import sentencepiece as spm
from sacrebleu import corpus_bleu
from tqdm import tqdm
import glob
import re
from model import Encoder, Decoder, PAD_ID

DEVICE = "cuda" if torch.cuda.is_available() else "cpu"
VOCAB = 16000

sp = spm.SentencePieceProcessor(model_file="bpe.model")
BOS = sp.bos_id()
EOS = sp.eos_id()

ckpts = glob.glob("checkpoint_epoch*.pt")
assert ckpts, "No checkpoint files found!"

ckpts = sorted(ckpts, key=lambda x: int(re.findall(r'\d+', x)[0]))
latest_ckpt = ckpts[-1]
print(f"Loading checkpoint: {latest_ckpt}")

checkpoint = torch.load(latest_ckpt, map_location=DEVICE)

enc = Encoder(VOCAB).to(DEVICE)
dec = Decoder(VOCAB).to(DEVICE)
enc.load_state_dict(checkpoint["enc"])
dec.load_state_dict(checkpoint["dec"])
enc.eval()
dec.eval()

def beam_search(src_ids, beam_size=4, max_len=100):
    src = torch.tensor([src_ids], dtype=torch.long).to(DEVICE)
    src_mask = (src != PAD_ID).float()

    with torch.no_grad():
        h, c = enc(src, src_mask)

    beams = [([BOS], 0.0, (h, c))]  # tokens, score, hidden state

    for _ in range(max_len):
        new_beams = []
        for tokens, score, (h, c) in beams:
            last = torch.tensor([[tokens[-1]]], dtype=torch.long).to(DEVICE)
            with torch.no_grad():
                logits, (h_new, c_new) = dec(last, (h, c))

            probs = torch.log_softmax(logits[:, -1, :], dim=-1).squeeze(0)
            topk = torch.topk(probs, beam_size)

            for tok, tok_score in zip(topk.indices.tolist(), topk.values.tolist()):
                if tok == EOS:
                    new_beams.append((tokens + [tok], score + tok_score, (h_new, c_new)))
                else:
                    new_beams.append((tokens + [tok], score + tok_score, (h_new, c_new)))

        beams = sorted(new_beams, key=lambda x: x[1], reverse=True)[:beam_size]

        if all(b[0][-1] == EOS for b in beams):
            break

    best = beams[0][0]
    return best[1:-1]  # removing BOS & EOS


test_src = open("test.src").read().splitlines()
test_tgt = open("test.tgt").read().splitlines()

print(f"Loaded {len(test_src)} test examples.")
references = []
hypotheses = []

print("Evaluating BLEU on test set...")

for s, t in tqdm(zip(test_src, test_tgt), total=len(test_src)):
    src_ids = list(map(int, s.split()))
    tgt_ids = list(map(int, t.split()))

    hyp_ids = beam_search(src_ids, beam_size=4)
    hyp = sp.decode(hyp_ids)

    ref = sp.decode(tgt_ids)

    hypotheses.append(hyp)
    references.append([ref])

bleu = corpus_bleu(hypotheses, references)
print("\n===== BLEU SCORE =====")
print(bleu)
```
